# ****CATATAN KOMPETISI****

---

****DESKRIPSI****

Machine Learning Solution Challenge merupakan kompetisi yang berfokus pada penerapan Machine Learning untuk menyelesaikan permasalahan nyata melalui pemanfaatan data secara inovatif, relevan, dan berdampak.

Sebagai bentuk dukungan terhadap pengembangan kompetensi mahasiswa, Enterprise Data Management (EDM) Laboratory menyelenggarakan kompetisi ini sebagai wadah untuk mendorong lahirnya solusi berbasis Artificial Intelligence, khususnya Machine Learning, yang selaras dengan salah satu tujuan SDGs. Melalui kompetisi ini, peserta diharapkan mampu menunjukkan kemampuan berpikir kritis, kolaborasi, komunikasi, serta penerapan teknologi yang sesuai dengan kebutuhan dunia industri.

Tema "Machine Learning Solution Challenge" mengajak peserta untuk mengembangan solusi berbasis Machine Learning dalam menjawab permasalahan nyata secara kreatif dan aplikatif.

**Setiap tim wajib memilih minimal satu dari 17 SDGs sebagai dasar dalam menentukan studi kasus yang akan diselesaikan. Dengan demikian, kompetisi ini diharapkan tidak hanya menjadi ajang untuk kemampuan teknis, tetapi juga sarana untuk menghasilkan solusi yang bermanfaat dan berkontribusi terhadap pencapaian SDGs.**

****KETENTUAN UMUM (PENTING)****

1.  Setiap tim wajib memili minimal satu dari 17 tujuan SDGs sebagai dasar dalam menentukan studi kasus yang akan diangkat.
2. Studi kasus yang dipilih harus relevan dengan permasalahan nyata dan dapat diselesaikan melalui pendekatan Machine Learning.
3. Solusi yang dikembangan harus menggunakan data yang relevan, valid, dan dapat dipertanggungjawabkan secara ilmiah.
4. Model Machine Learning yang dibangun harus disesuaikan dengan karakteristik data dan permasalahan yang diangkat.
5. Peserta wajib melakukan pengolahan data, analisis, pemodelan, dan evaluasi secara sistematis.
6. Peserta wajib mencamtumkan sumber data serta seluruh referensi dan pustaka yang digunakan dalam daftar pustaka pada materi presentasi.

****PENYISIHAN****

Babak penyisihan merupakan tahap awal dalam Machine Learning Solution Challenge. Pada tahap ini, seluruh tim peserta akan mengembangkan solusi berbasis Machine Learning sesuai dengan tema kompetisi. Setiap tim wajib memilih minimal satu tujuan SDGs sebagai dasar studi kasus. Peserta diwajibkan melakukan analisis permasalahan, pengolahan data, pembangunan model Machine Learning, evaluasi performa model, serta mengumpulkan hasil pengembangan berupa:

1. Source Code
2. Dataset

****PENGEMBANGAN SOLUSI****

1. Peserta mengembangkan solusi Machine Learning sesuai tema kompetisi.
2. Peserta melakukan pengumpulan data, preprocessing, pembangunan model, dan evaluasi.
3. Seluruh hasil pekerjaan wajib dikumpulkan melalui media pengumpulan yang disediakan oleh panitia.
4. Dilaksanakan pada 13 - 28 November 2026.

****PENGATURAN UMUM PENYISIHAN****

1. Dataset yang digunakan dapat berasal dari sumber terbuka maupun dataset yang dikumpulkan sendiri dengan tetap mencamtumkan sumber data pada daftar pustaka dalam materi presentasi.
2. Solusi yang dikembangkan wajib menggunakan metode Machine Learning.
3. Framework atau bahasa pemrograan yang digunakan tidak dibatasi (Python, R, Julia, dll).
4. Library yang digunakan bebas, seperti Scikit-Learn, TensorFlow, PyTorch, XGBoost, CatBoost, LightGBM, dan library pendukung lainnya.
5. Setiap tim wajib menyerahkan:
    - Source Code
    - Dataset

Penilaian tidak hanya berfokus pada performa model, tetapi juga kemampuan peserta dalam menjelaskan metodologi, hasil analisis, serta dampak solusi yang dihasilkan.

****KETERANGAN PENILAIAN****

1. Problem Identification (Perumusan Permasalahan), menilai kemampuan peserta dalam mengidentifikasi permasalahan nyata yang relevan dengan SDGs serta merumuskan tujuan penyelesaian yang jelas.

2. Data  Preparation  (Pengolahan  dan  Kualitas  Data),  menilai  kualitas  dataset yang digunakan, proses preprocessing, serta kesiapan data sebelum digunakan dalam pembangunan model Machine Learning.

3. Machine Learning Implementation (Implementasi Machine Learning), menilai ketepatan pemilihan algoritma, proses pembangunan model, dan kesesuaian metode yang digunakan dengan permasalahan yang diangkat.

4. Model Evaluation (Evaluasi Model), menilai performa model berdasarkan metrik evaluasi yang digunakan serta kemampuan peserta dalam menginterpretasikan hasil yang diperoleh.

5. Innovation (Inovasi dan Dampak Solusi), menilai kreativitas, kebaruan ide, dan nilai tambah solusi yang dikembangkan untuk memberikan dampak terhadap permasalahan yang dipilih
